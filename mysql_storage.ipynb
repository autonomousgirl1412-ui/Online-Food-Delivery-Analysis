```python
import pandas as pd
from sqlalchemy import create_engine, text

# 1. Load final feature-engineered data
df = pd.read_csv("03_final_data.csv")
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
print("Data loaded successfully! Shape:", df.shape)

# 2. Create database if not exists
base_engine = create_engine("mysql+pymysql://root:vish1412@localhost")
with base_engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS food_delivery_db;"))
print("Database 'food_delivery_db' verified!")

# 3. Insert 100,000 rows in chunks
engine = create_engine("mysql+pymysql://root:vish1412@localhost/food_delivery_db")
df.to_sql(
    "food_delivery_orders",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=5000
)

# 4. Verify row count in MySQL
with engine.connect() as conn:
    count = conn.execute(text("SELECT COUNT(*) FROM food_delivery_orders;")).scalar()
print(f"Success! Total rows verified in MySQL: {count}")
```
